```python
import pandas as pd
import numpy as np

raw_df = pd.read_csv('indian_toy_market_sales_data.csv')

df = raw_df.copy()
df.head()
df.duplicated().sum()
df.drop_duplicates()
```

```python
df.columns = df.columns.str.strip().str.lower()
```

```python
df['transaction_date'] = pd.to_datetime(df['transaction_date'], format='mixed', errors='coerce') 
```

```python
df = df.drop(columns=['year','month','month_name'])
```

```python
df['year'] = df['transaction_date'].dt.year
df['month'] = df['transaction_date'].dt.month
df['month_name'] = df['transaction_date'].dt.month_name()
```

```python
df['month_name'] = df['month_name'].str.strip().str.capitalize()
df['month_name'].unique()
```

```python
df['quarter'].unique()
```

```python
df = df.drop(columns=['day_of_week'])
```

```python
df['day_of_week'] = df['transaction_date'].dt.day_name()
```

```python
df['day_of_week'] = df['day_of_week'].str.strip().str.capitalize()
df['day_of_week'].unique()
```

```python
df['brand'] = df['brand'].str.strip().str.lower().replace( {
    'okplay': 'OK Play',
    'ok play': 'OK Play',
    'lego': 'LEGO',
    'funskool india': 'Funskool',
    'funskool': 'Funskool',
    'hasbro': 'Hasbro',
    'fisher price': 'Fisher-Price',
    'fisher-price': 'Fisher-Price',
    'nerf': 'NERF',
    'hotwheels': 'Hot Wheels',
    'hot wheels': 'Hot Wheels',
    'toyzone': 'Toyzone',
    'toy zone': 'Toyzone',
    'barbie': 'Barbie',
    'skillmatics': 'Skillmatics',
    'playshifu': 'PlayShifu',
    'mattel': 'Mattel',
    'smartivity': 'Smartivity'
})

df['brand'].unique()
```

```python
df['brand_type'].unique()
```

```python
df['product_category'] = df['product_category'].str.strip().str.title()
df['product_category'].unique()
```

```python
df['product_subcategory'] = df['product_subcategory'].str.strip()
df['product_subcategory'].unique()
```

```python
df['region'].unique()
```

```python
df['city'] = df['city'].str.strip().str.title().replace({
    'Calcutta': 'Kolkata',
    'Bombay': 'Mumbai',
    'Bangalore': 'Bengaluru'
})

df['city'].unique()
```

```python
df['city_tier'].unique()
```

```python
df['sales_channel'] = df['sales_channel'].str.strip().str.lower().replace({'hamleys store':'hamleys',
                                                                               'website':'brand website',
                                                                               'amazon.in':'amazon'}).str.title()

df['sales_channel'].unique()
```

```python
df['channel_type'].unique()
```

```python
df['target_age_group'] = df['target_age_group'].str.strip().replace({'13+':'13+ years',
                                                                     '9-12':'9-12 years',
                                                                     '0-2':'0-2 years',
                                                                     '3-5':'3-5 years', 
                                                                     '6-8':'6-8 years'})
df['target_age_group'].unique()
```

```python
df.loc[(df['unit_price_inr'] < 0),'unit_price_inr'] = np.nan
df['unit_price_inr'].unique()
(df['unit_price_inr'] < 0).any()
```

```python
df.loc[(df['quantity'] < 0),'quantity'] = np.nan
df['quantity'] = df['quantity'].astype('Int64')
df['quantity'].unique()
```

```python
df = df.drop(columns=['gross_revenue_inr'])
```

```python
df.loc[(df['discount_percentage'] < 0) | (df['discount_percentage'] > 100),'discount_percentage'] = np.nan
df['discount_percentage'].unique()
```

```python
df = df.drop(columns=['discount_amount_inr'])

df = df.drop(columns=['net_revenue_inr'])
```

```python
df.loc[(df['customer_rating'] < 0)|(df['customer_rating'] > 5 ),'customer_rating'] = np.nan

df['customer_rating'].unique()
```

```python
df['safety_certified'] = df['safety_certified'].str.strip().replace({'TRUE':'Yes',
                                                                     'FALSE':'No',
                                                                     'no':'No',
                                                                     'yes':'Yes',
                                                                     'N':'No', 
                                                                     'Y':'Yes'})

df['safety_certified'].unique()
```

```python
df['returned'] = df['returned'].str.strip().replace({'TRUE':'Yes',
                                                    'FALSE':'No',
                                                     'no':'No',
                                                    'yes':'Yes',
                                                    'N':'No', 
                                                    'Y':'Yes'})
df['returned'].unique()
```

```python
df['customer_type'].unique()
```

```python
df.dtypes
```

```python
df['region'] = df.groupby(['city'])['region'].transform(lambda x: x.fillna(x.mode()[0])) 
```

```python
df['unit_price_inr'] = df.groupby(['product_subcategory','brand','brand_type'])['unit_price_inr'].transform(lambda x: x.fillna(x.median()))
```

```python
df['quantity'] = df.groupby(['product_subcategory','product_category'])['quantity'].transform(lambda x: x.fillna(x.median()))
```

```python
df['customer_rating'] = df.groupby(['brand','brand_type','product_category','product_subcategory'])['customer_rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df['safety_certified'] = df.groupby(['brand','brand_type','product_category','product_subcategory'])['safety_certified'].transform(lambda x: x.fillna(x.mode()[0]))
```

```python
df['customer_type'] = df.groupby(['city_tier','sales_channel','channel_type','target_age_group','region'])['customer_type'].transform(lambda x: x.fillna(x.mode()[0]))
```

```python
df['returned'] = df.groupby(['brand'])['returned'].transform(lambda x: x.fillna(x.mode()[0]))
```

```python
df['net_revenue_inr'] = df['quantity'] * df['unit_price_inr'] * (1 - df['discount_percentage']/100)
```

```python
(df['net_revenue_inr'] != df['quantity'] * df['unit_price_inr'] * (1 - df['discount_percentage']/100)).any()
```
